## **3. Growth Potential**



*   Feature engineering of Growth Potential and player classification into categories (World Class, Excellent, Good, Average).



```python
# Growth Potential
df["Growth Potential"] = df["Potential"] - df["Overall"]

def rating_category(rating):
    if rating >= 90:
        return "World Class"
    elif rating >= 85:
        return "Excellent"
    elif rating >= 80:
        return "Good"
    else:
        return "Average"

df["Rating Category"] = df["Overall"].apply(rating_category)
df.head()
```

# **4. Correlation & Heatmap**




*  Visualizing relationships between key metrics like Age, Rating, Potential, and technical skills



```python
numeric_cols = [
    "Age", "Overall", "Potential", "Growth Potential",
    "Weak Foot", "Skill Moves", "Strength", "Stamina",
    "ShotPower", "Penalties"
]
# Filter to existing numeric columns
numeric_cols = [col for col in numeric_cols if col in df.columns]

df_corr = df[numeric_cols].corr()
df_corr

plt.figure(figsize=(10, 8))
sns.heatmap(df_corr, cmap='viridis', annot=True, fmt='.2f')
plt.title("Correlation Heatmap of FIFA Player Attributes")
plt.show()
```

# **5. Histogram**

# **12. Additional FIFA Analysis**

```python
# Define the top 10 players by Overall and Potential to fix the NameError in subsequent cells
top_players = df.nlargest(10, 'Overall')
top_potential = df.nlargest(10, 'Potential')
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    top_players['Overall'],
    labels=top_players['Name'],
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('viridis', len(top_players))
)
plt.title("Top 10 Highest Rated FIFA Players (Rating Proportion)")
plt.show()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    top_potential['Potential'],
    labels=top_potential['Name'],
    autopct='%1.1f%%',
    startangle=140,
    colors=sns.color_palette('magma', len(top_potential))
)
plt.title("Top 10 Players by Potential (Proportion)")
plt.show()
```

```python
# Scatterplots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.scatterplot(data=df, x="Age", y="Overall", ax=axes[0], alpha=0.6, color='royalblue')
axes[0].set_title("Age vs Overall Rating")
sns.scatterplot(data=df, x="Overall", y="Potential", ax=axes[1], alpha=0.6, color='darkorange')
axes[1].set_title("Overall Rating vs Potential")
plt.tight_layout()
plt.show()
```

```python
# Line Charts
age_overall = df.groupby('Age')['Overall'].mean().reset_index()
age_growth = df.groupby('Age')['Growth Potential'].mean().reset_index()

plt.figure(figsize=(10, 5))
plt.plot(age_overall['Age'], age_overall['Overall'], color='royalblue', marker='o', linewidth=2, label='Avg Overall')
plt.title('Average Overall Rating by Age')
plt.xlabel('Age')
plt.ylabel('Overall Rating')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(age_growth['Age'], age_growth['Growth Potential'], color='crimson', marker='s', linewidth=2, label='Avg Growth Potential')
plt.title('Average Growth Potential by Age')
plt.xlabel('Age')
plt.ylabel('Growth Potential')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
